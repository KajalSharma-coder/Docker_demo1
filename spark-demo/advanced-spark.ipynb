```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
```

```python
spark = SparkSession.builder \
    .appName("AdvanceSparkApp") \
    .getOrCreate()

spark
```

```python
df = spark.range(0, 1000000) \
    .withColumn("value", col("id") * 1000)

df.show(5)
```

```python
df.printSchema()
```

```python
df.count()
```

```python
filtered_df = df.filter(col("id") > 500000)

filtered_df.show(10)
```

```python
new_df = df.withColumn("salary", col("id") * 500)

new_df.show(5)
```
